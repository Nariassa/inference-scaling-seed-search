```python
import os
import json

TARGET = "/content/drive/MyDrive/samples"
PROMPT_FILE = "/content/inference-scaling-seed-search/data/geneval_subset_120.jsonl"

with open(PROMPT_FILE, "r", encoding="utf-8") as f:
    prompts = [json.loads(line) for line in f if line.strip()]

# 建立 prompt_id -> prompt 的映射
prompt_dict = {
    int(item["prompt_id"]): item["prompt"]
    for item in prompts
}

valid_images = 0
missing_images = []
mapping_examples = []

for prompt_id in sorted(prompt_dict):
    for seed in range(32):
        image_number = prompt_id * 32 + seed

        image_path = os.path.join(
            TARGET,
            f"{prompt_id:05d}",
            f"{image_number:05d}.png"
        )

        if os.path.exists(image_path):
            valid_images += 1
        else:
            missing_images.append(image_path)

        if prompt_id < 2 and seed < 3:
            mapping_examples.append({
                "prompt_id": prompt_id,
                "seed": seed,
                "image_path": image_path,
                "prompt": prompt_dict[prompt_id]
            })

print("========== Mapping validation ==========")
print("Total prompts:", len(prompt_dict))
print("Expected images:", len(prompt_dict) * 32)
print("Valid images:", valid_images)
print("Missing images:", len(missing_images))

print("\n========== Mapping examples ==========")
for item in mapping_examples:
    print(item)

if missing_images:
    print("\nFirst missing paths:")
    for path in missing_images[:10]:
        print(path)
else:
    print("\nSUCCESS: All image paths are valid!")
```

```python
%%writefile /content/inference-scaling-seed-search/eval/clip_scorer.py

import os
import json
import argparse
import pandas as pd
import torch
import open_clip

from PIL import Image
from tqdm.auto import tqdm


def main():
    parser = argparse.ArgumentParser()

    parser.add_argument(
        "--repo_dir",
        default="/content/inference-scaling-seed-search"
    )

    parser.add_argument(
        "--image_dir",
        default="/content/drive/MyDrive/samples"
    )

    parser.add_argument(
        "--batch_size",
        type=int,
        default=64
    )

    args = parser.parse_args()

    # --------------------------------------------------
    # 1. Paths
    # --------------------------------------------------

    repo_dir = args.repo_dir
    image_dir = args.image_dir

    prompt_file = os.path.join(
        repo_dir, "data", "geneval_subset_120.jsonl"
    )

    output_dir = os.path.join(repo_dir, "results")
    os.makedirs(output_dir, exist_ok=True)

    output_file = os.path.join(
        output_dir, "clip_scores.csv"
    )

    # --------------------------------------------------
    # 2. Device
    # --------------------------------------------------

    device = "cuda" if torch.cuda.is_available() else "cpu"

    print("=" * 60)
    print("CLIP Scoring")
    print("=" * 60)
    print("Device:", device)

    if device == "cuda":
        print("GPU:", torch.cuda.get_device_name(0))

    # --------------------------------------------------
    # 3. Load prompts
    # --------------------------------------------------

    prompts = []

    with open(prompt_file, "r", encoding="utf-8") as f:
        for line in f:
            if line.strip():
                prompts.append(json.loads(line))

    prompt_dict = {
        int(item["prompt_id"]): item
        for item in prompts
    }

    print("Number of prompts:", len(prompt_dict))

    # --------------------------------------------------
    # 4. Load CLIP model
    # --------------------------------------------------

    print("\nLoading OpenAI CLIP ViT-B/32...")

    model, _, preprocess = open_clip.create_model_and_transforms(
        "ViT-B-32",
        pretrained="openai"
    )

    tokenizer = open_clip.get_tokenizer("ViT-B-32")

    model = model.to(device)
    model.eval()

    print("CLIP model loaded successfully.")

    # --------------------------------------------------
    # 5. Pre-compute text embeddings
    # --------------------------------------------------

    prompt_ids = sorted(prompt_dict.keys())

    prompt_texts = [
        prompt_dict[pid]["prompt"]
        for pid in prompt_ids
    ]

    text_tokens = tokenizer(prompt_texts).to(device)

    with torch.no_grad():
        text_features = model.encode_text(text_tokens)
        text_features = text_features / text_features.norm(
            dim=-1, keepdim=True
        )

    prompt_to_index = {
        pid: index
        for index, pid in enumerate(prompt_ids)
    }

    print("Text embeddings computed:", len(prompt_texts))

    # --------------------------------------------------
    # 6. Resume support
    # --------------------------------------------------

    completed = set()

    if os.path.exists(output_file):
        old_df = pd.read_csv(output_file)

        completed = set(
            zip(
                old_df["prompt_id"].astype(int),
                old_df["seed"].astype(int)
            )
        )

        print("Existing results:", len(completed))
        print("Resuming unfinished images...")

    # --------------------------------------------------
    # 7. Build image records
    # --------------------------------------------------

    records = []

    for prompt_id in prompt_ids:

        prompt_info = prompt_dict[prompt_id]

        for local_seed_idx in range(32):

            # Global seed used by the project
            seed = prompt_id * 32 + local_seed_idx

            if (prompt_id, seed) in completed:
                continue

            image_path = os.path.join(
                image_dir,
                f"{prompt_id:05d}",
                f"{seed:05d}.png"
            )

            if not os.path.exists(image_path):
                print("WARNING: Missing image:", image_path)
                continue

            records.append({
                "prompt_id": prompt_id,
                "seed": seed,
                "local_seed_idx": local_seed_idx,
                "prompt": prompt_info["prompt"],
                "tag": prompt_info.get("tag"),
                "image_path": image_path,
                "prompt_index": prompt_to_index[prompt_id]
            })

    print("\nImages remaining:", len(records))

    # --------------------------------------------------
    # 8. Batch image scoring
    # --------------------------------------------------

    batch_size = args.batch_size

    for start in tqdm(
        range(0, len(records), batch_size),
        desc="CLIP scoring"
    ):

        batch_records = records[start:start + batch_size]

        images = []
        valid_records = []

        for record in batch_records:
            try:
                image = Image.open(
                    record["image_path"]
                ).convert("RGB")

                images.append(preprocess(image))
                valid_records.append(record)

            except Exception as e:
                print(
                    "\nImage loading failed:",
                    record["image_path"],
                    str(e)
                )

        if not images:
            continue

        image_tensor = torch.stack(images).to(device)

        prompt_indices = torch.tensor(
            [r["prompt_index"] for r in valid_records],
            dtype=torch.long,
            device=device
        )

        with torch.no_grad():

            if device == "cuda":
                with torch.autocast(
                    device_type="cuda",
                    dtype=torch.float16
                ):
                    image_features = model.encode_image(image_tensor)
            else:
                image_features = model.encode_image(image_tensor)

            image_features = image_features.float()

            image_features = image_features / image_features.norm(
                dim=-1,
                keepdim=True
            )

            corresponding_text_features = text_features[
                prompt_indices
            ].float()

            similarities = (
                image_features * corresponding_text_features
            ).sum(dim=-1)

        # --------------------------------------------------
        # 9. Save current batch
        # --------------------------------------------------

        batch_results = []

        for record, score in zip(
            valid_records,
            similarities.cpu().tolist()
        ):

            batch_results.append({
                "prompt_id": record["prompt_id"],
                "seed": record["seed"],
                "local_seed_idx": record["local_seed_idx"],
                "prompt": record["prompt"],
                "tag": record["tag"],
                "image_path": record["image_path"],
                "clip_cosine_similarity": score
            })

        batch_df = pd.DataFrame(batch_results)

        batch_df.to_csv(
            output_file,
            mode="a",
            header=not os.path.exists(output_file),
            index=False
        )

    # --------------------------------------------------
    # 10. Final summary
    # --------------------------------------------------

    print("\n" + "=" * 60)
    print("CLIP SCORING COMPLETED")
    print("=" * 60)

    if os.path.exists(output_file):

        final_df = pd.read_csv(output_file)

        print("Output:", output_file)
        print("Total scored images:", len(final_df))
        print(
            "Unique prompts:",
            final_df["prompt_id"].nunique()
        )
        print(
            "Unique image keys:",
            final_df[["prompt_id", "seed"]].drop_duplicates().shape[0]
        )

        print("\nCLIP score statistics:")
        print(
            final_df["clip_cosine_similarity"].describe()
        )

        print("\nFirst 5 results:")
        print(final_df.head().to_string(index=False))


if __name__ == "__main__":
    main()
```

```python

%%writefile /content/inference-scaling-seed-search/eval/diversity_scorer.py

import os
import numpy as np
import pandas as pd
import torch
import open_clip

from PIL import Image
from tqdm.auto import tqdm

# ============================================================
# Configuration
# ============================================================

REPO_DIR = "/content/inference-scaling-seed-search"
INPUT_CSV = os.path.join(
    REPO_DIR, "results", "clip_scores_quickgelu.csv"
)

OUTPUT_CSV = os.path.join(
    REPO_DIR, "results", "clip_diversity_per_prompt.csv"
)

TAG_OUTPUT_CSV = os.path.join(
    REPO_DIR, "results", "clip_diversity_by_tag.csv"
)

BATCH_SIZE = 64
MODEL_NAME = "ViT-B-32-quickgelu"

device = "cuda" if torch.cuda.is_available() else "cpu"

print("=" * 60)
print("CLIP Diversity Evaluation")
print("=" * 60)
print("Device:", device)

if device == "cuda":
    print("GPU:", torch.cuda.get_device_name(0))

# ============================================================
# Load image records
# ============================================================

df = pd.read_csv(INPUT_CSV)

required_columns = [
    "prompt_id",
    "prompt",
    "tag",
    "local_seed_idx",
    "image_path"
]

for col in required_columns:
    if col not in df.columns:
        raise ValueError(f"Missing required column: {col}")

df = df.sort_values(
    ["prompt_id", "local_seed_idx"]
).reset_index(drop=True)

assert len(df) == 3840, f"Expected 3840 images, got {len(df)}"

print("Total images:", len(df))
print("Unique prompts:", df["prompt_id"].nunique())

# ============================================================
# Load the same CLIP model
# ============================================================

print("\nLoading CLIP ViT-B-32-quickgelu...")

model, _, preprocess = open_clip.create_model_and_transforms(
    MODEL_NAME,
    pretrained="openai",
    device=device
)

model.eval()

print("CLIP model loaded successfully.")

# ============================================================
# Extract normalized image embeddings
# ============================================================

all_embeddings = []

paths = df["image_path"].tolist()

with torch.no_grad():

    for start in tqdm(
        range(0, len(paths), BATCH_SIZE),
        desc="Extracting image embeddings"
    ):

        batch_paths = paths[start:start + BATCH_SIZE]

        images = []

        for path in batch_paths:
            if not os.path.exists(path):
                raise FileNotFoundError(path)

            image = Image.open(path).convert("RGB")
            images.append(preprocess(image))

        image_tensor = torch.stack(images).to(device)

        features = model.encode_image(image_tensor)

        features = features / features.norm(
            dim=-1, keepdim=True
        )

        all_embeddings.append(
            features.cpu().numpy()
        )

        del image_tensor, features, images

embeddings = np.concatenate(all_embeddings, axis=0)

print("\nEmbedding shape:", embeddings.shape)

# ============================================================
# Calculate pairwise CLIP embedding diversity
# ============================================================

results = []

for prompt_id, group in df.groupby("prompt_id", sort=True):

    indices = group.index.to_numpy()
    prompt_embeddings = embeddings[indices]

    if len(prompt_embeddings) != 32:
        raise ValueError(
            f"Prompt {prompt_id} has {len(prompt_embeddings)} images, not 32"
        )

    # Embeddings are normalized, so dot product = cosine similarity
    similarity_matrix = (
        prompt_embeddings @ prompt_embeddings.T
    )

    upper_indices = np.triu_indices(32, k=1)

    pairwise_similarity = similarity_matrix[upper_indices]

    pairwise_distance = 1.0 - pairwise_similarity

    results.append({
        "prompt_id": int(prompt_id),
        "prompt": group["prompt"].iloc[0],
        "tag": group["tag"].iloc[0],
        "num_images": len(group),
        "num_pairs": len(pairwise_distance),
        "mean_pairwise_cosine_similarity": float(
            pairwise_similarity.mean()
        ),
        "clip_embedding_diversity": float(
            pairwise_distance.mean()
        ),
        "std_pairwise_cosine_distance": float(
            pairwise_distance.std()
        )
    })

result_df = pd.DataFrame(results)

result_df.to_csv(OUTPUT_CSV, index=False)

# ============================================================
# Aggregate by GenEval category
# ============================================================

tag_df = (
    result_df.groupby("tag")
    .agg(
        num_prompts=("prompt_id", "count"),
        mean_diversity=("clip_embedding_diversity", "mean"),
        std_diversity=("clip_embedding_diversity", "std"),
        mean_similarity=("mean_pairwise_cosine_similarity", "mean")
    )
    .reset_index()
)

tag_df.to_csv(TAG_OUTPUT_CSV, index=False)

# ============================================================
# Summary
# ============================================================

print("\n" + "=" * 60)
print("DIVERSITY EVALUATION COMPLETED")
print("=" * 60)

print("Total prompts:", len(result_df))
print("Images per prompt:", result_df["num_images"].unique())
print("Total image pairs:", result_df["num_pairs"].sum())

print("\nOverall diversity statistics:")
print(
    result_df["clip_embedding_diversity"].describe()
)

print("\nDiversity by tag:")
print(tag_df.to_string(index=False))

print("\nOutput files:")
print(OUTPUT_CSV)
print(TAG_OUTPUT_CSV)
```

```python

%%writefile /content/inference-scaling-seed-search/eval/matched_compute.py

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# Configuration
# ============================================================

REPO = "/content/inference-scaling-seed-search"

BON_SWEEP = os.path.join(
    REPO, "results/results_bon/bon_sweep.csv"
)

BON_SELECTIONS = os.path.join(
    REPO, "results/results_bon/bon_selections.csv"
)

GENEVAL_FILE = os.path.join(
    REPO, "data/geneval_per_image.csv"
)

PSP_SUMMARY = os.path.join(
    REPO, "results/results_psp/budget/summary.csv"
)

PSP_GROUPS = os.path.join(
    REPO, "results/results_psp/budget/per_group.csv"
)

OUTPUT_DIR = os.path.join(
    REPO, "results"
)

os.makedirs(OUTPUT_DIR, exist_ok=True)

MATCHED_BUDGETS = [128, 256, 512, 1024]

# ============================================================
# 1. Load existing results
# ============================================================

print("=" * 70)
print("MATCHED-COMPUTE ANALYSIS")
print("=" * 70)

bon_sweep = pd.read_csv(BON_SWEEP)
bon = pd.read_csv(BON_SELECTIONS)
geneval = pd.read_csv(GENEVAL_FILE)
psp = pd.read_csv(PSP_SUMMARY)
psp_groups = pd.read_csv(PSP_GROUPS)

print("BoN sweep rows:", len(bon_sweep))
print("BoN selection rows:", len(bon))
print("GenEval rows:", len(geneval))
print("PSP configurations:", len(psp))
print("PSP group rows:", len(psp_groups))

# ============================================================
# 2. Evaluate BoN selected images with GenEval
# ============================================================

print("\n[1/5] Joining BoN selections with GenEval...")

bon_eval = bon.merge(
    geneval[
        [
            "prompt_id",
            "seed",
            "correct",
            "geneval_soft_score"
        ]
    ],
    on=["prompt_id", "seed"],
    how="left",
    validate="many_to_one",
    indicator=True
)

if (bon_eval["_merge"] != "both").any():
    missing = (bon_eval["_merge"] != "both").sum()
    raise ValueError(
        f"{missing} BoN selections could not match GenEval."
    )

bon_eval = bon_eval.drop(columns=["_merge"])

print("All BoN selections matched successfully.")

# ============================================================
# 3. Calculate BoN group-level metrics
# ============================================================

print("\n[2/5] Calculating BoN group metrics...")

bon_group = (
    bon_eval.groupby(["N", "group"])
    .agg(
        num_prompts=("prompt_id", "nunique"),
        image_reward_mean=("image_reward", "mean"),
        geneval_mean=("correct", "mean"),
        geneval_soft_mean=("geneval_soft_score", "mean")
    )
    .reset_index()
)

bon_group["compute_C"] = bon_group["N"] * 64
bon_group["method"] = "BoN"
bon_group["configuration"] = (
    "bon_n" + bon_group["N"].astype(str)
)

# Aggregate group means, giving each group equal weight
bon_summary = (
    bon_group.groupby(["N", "compute_C"])
    .agg(
        image_reward_mean=("image_reward_mean", "mean"),
        image_reward_std=("image_reward_mean", "std"),
        geneval_mean=("geneval_mean", "mean"),
        geneval_std=("geneval_mean", "std"),
        geneval_soft_mean=("geneval_soft_mean", "mean"),
        num_groups=("group", "nunique"),
        num_prompts=("num_prompts", "sum")
    )
    .reset_index()
)

bon_summary["method"] = "BoN"
bon_summary["configuration"] = (
    "bon_n" + bon_summary["N"].astype(str)
)

# Verify that the reconstructed IR agrees with A's sweep
verification = bon_summary.merge(
    bon_sweep[["N", "ir_mean"]],
    on="N",
    how="left"
)

verification["ir_abs_difference"] = (
    verification["image_reward_mean"]
    - verification["ir_mean"]
).abs()

print("\nBoN ImageReward verification:")
print(
    verification[
        ["N", "compute_C", "image_reward_mean",
         "ir_mean", "ir_abs_difference"]
    ].to_string(index=False)
)

if verification["ir_abs_difference"].max() > 1e-5:
    raise ValueError(
        "Reconstructed BoN ImageReward does not match bon_sweep.csv."
    )

print("BoN ImageReward verification passed.")

# ============================================================
# 4. Prepare PSP results
# ============================================================

print("\n[3/5] Preparing PSP results...")

psp_matched = psp[
    psp["compute_C"].isin(MATCHED_BUDGETS)
].copy()

psp_matched["N"] = psp_matched["nbar"]
psp_matched["method"] = "PSP"

psp_matched = psp_matched.rename(
    columns={
        "image_reward_mean": "image_reward_mean",
        "image_reward_std": "image_reward_std",
        "geneval_mean": "geneval_mean",
        "geneval_std": "geneval_std"
    }
)

psp_matched["geneval_soft_mean"] = np.nan
psp_matched["num_groups"] = psp_matched["num_groups"]
psp_matched["num_prompts"] = psp_matched["num_prompts"]

# Keep all PSP configurations
psp_output = psp_matched[
    [
        "method",
        "configuration",
        "compute_C",
        "N",
        "image_reward_mean",
        "image_reward_std",
        "geneval_mean",
        "geneval_std",
        "geneval_soft_mean",
        "num_groups",
        "num_prompts"
    ]
].copy()

bon_output = bon_summary[
    [
        "method",
        "configuration",
        "compute_C",
        "N",
        "image_reward_mean",
        "image_reward_std",
        "geneval_mean",
        "geneval_std",
        "geneval_soft_mean",
        "num_groups",
        "num_prompts"
    ]
].copy()

all_results = pd.concat(
    [bon_output, psp_output],
    ignore_index=True
)

all_results = all_results.sort_values(
    ["compute_C", "method", "configuration"]
).reset_index(drop=True)

# ============================================================
# 5. Save all comparison tables
# ============================================================

print("\n[4/5] Saving comparison tables...")

all_path = os.path.join(
    OUTPUT_DIR, "matched_compute_all.csv"
)

group_path = os.path.join(
    OUTPUT_DIR, "matched_compute_group_metrics.csv"
)

best_path = os.path.join(
    OUTPUT_DIR, "matched_compute_best_psp_descriptive.csv"
)

all_results.to_csv(all_path, index=False)

group_output = bon_group.copy()

psp_group_matched = psp_groups[
    psp_groups["compute_C"].isin(MATCHED_BUDGETS)
].copy()

psp_group_matched["method"] = "PSP"

group_output.to_csv(group_path, index=False)

psp_group_path = os.path.join(
    OUTPUT_DIR, "matched_compute_psp_group_metrics.csv"
)

psp_group_matched.to_csv(
    psp_group_path, index=False
)

best_psp = (
    psp_matched.sort_values(
        ["compute_C", "image_reward_mean"],
        ascending=[True, False]
    )
    .groupby("compute_C")
    .head(1)
    .copy()
)

best_psp.to_csv(best_path, index=False)

# ============================================================
# 6. Plot ImageReward comparison
# ============================================================

print("\n[5/5] Creating plots...")

fig, ax = plt.subplots(figsize=(9, 5))

bon_plot = bon_summary.sort_values("compute_C")

ax.plot(
    bon_plot["compute_C"],
    bon_plot["image_reward_mean"],
    marker="o",
    linewidth=2,
    label="BoN"
)

for _, row in psp_matched.iterrows():
    ax.scatter(
        row["compute_C"],
        row["image_reward_mean"],
        marker="x",
        s=65
    )

# Add one legend entry for PSP configurations
ax.scatter(
    [],
    [],
    marker="x",
    label="PSP configurations"
)

ax.set_xlabel("Compute C (denoising trajectory-steps)")
ax.set_ylabel("Mean ImageReward")
ax.set_title("BoN vs PSP under Matched Compute")
ax.set_xticks(MATCHED_BUDGETS)
ax.grid(True, alpha=0.3)
ax.legend()

fig.tight_layout()

ir_plot_path = os.path.join(
    OUTPUT_DIR, "matched_compute_imagereward.png"
)

fig.savefig(ir_plot_path, dpi=200)
plt.show()
plt.close(fig)

# ============================================================
# 7. Plot GenEval comparison
# ============================================================

fig, ax = plt.subplots(figsize=(9, 5))

ax.plot(
    bon_plot["compute_C"],
    bon_plot["geneval_mean"],
    marker="o",
    linewidth=2,
    label="BoN"
)

for _, row in psp_matched.iterrows():
    ax.scatter(
        row["compute_C"],
        row["geneval_mean"],
        marker="x",
        s=65
    )

ax.scatter(
    [],
    [],
    marker="x",
    label="PSP configurations"
)

ax.set_xlabel("Compute C (denoising trajectory-steps)")
ax.set_ylabel("GenEval accuracy")
ax.set_title("GenEval under Matched Compute")
ax.set_xticks(MATCHED_BUDGETS)
ax.set_ylim(0, 1)
ax.grid(True, alpha=0.3)
ax.legend()

fig.tight_layout()

geneval_plot_path = os.path.join(
    OUTPUT_DIR, "matched_compute_geneval.png"
)

fig.savefig(geneval_plot_path, dpi=200)
plt.show()
plt.close(fig)

# ============================================================
# 8. Final report
# ============================================================

print("\n" + "=" * 70)
print("MATCHED-COMPUTE ANALYSIS COMPLETED")
print("=" * 70)

print("\nAll matched results:")
print(
    all_results[
        [
            "method",
            "configuration",
            "compute_C",
            "image_reward_mean",
            "geneval_mean"
        ]
    ].to_string(index=False)
)

print("\nDescriptive highest-ImageReward PSP configurations:")
print(
    best_psp[
        [
            "compute_C",
            "configuration",
            "image_reward_mean",
            "geneval_mean"
        ]
    ].to_string(index=False)
)

print("\nOutput files:")
for path in [
    all_path,
    group_path,
    psp_group_path,
    best_path,
    ir_plot_path,
    geneval_plot_path
]:
    print(path)
```

```python
%%writefile /content/inference-scaling-seed-search/eval/verifier_ablation.py

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

REPO = "/content/inference-scaling-seed-search"
RESULTS = os.path.join(REPO, "results")
os.makedirs(RESULTS, exist_ok=True)

# --------------------------------------------------
# 1. Load data
# --------------------------------------------------

print("=" * 65)
print("FINAL-CANDIDATE VERIFIER ABLATION")
print("=" * 65)

cache_path = os.path.join(REPO, "cache/T64_metrics.csv")
clip_path = os.path.join(RESULTS, "clip_scores_quickgelu.csv")
geneval_path = os.path.join(REPO, "data/geneval_per_image.csv")

cache = pd.read_csv(cache_path)
clip = pd.read_csv(clip_path)
geneval = pd.read_csv(geneval_path)

# Only use final ImageReward scores (step = 64)
final_ir = cache[cache["step"] == 64].copy()

# Reconstruct local seed index using the same ordering as BoN
final_ir = final_ir.sort_values(
    ["prompt_id", "seed"]
).reset_index(drop=True)

final_ir["local_seed_idx"] = (
    final_ir.groupby("prompt_id").cumcount()
)

print("Final ImageReward rows:", len(final_ir))
print("CLIP rows:", len(clip))
print("GenEval rows:", len(geneval))

# --------------------------------------------------
# 2. Merge all metrics
# --------------------------------------------------

df = final_ir[
    ["prompt_id", "seed", "local_seed_idx", "image_reward"]
].merge(
    clip[
        ["prompt_id", "seed", "local_seed_idx",
         "clip_cosine_similarity"]
    ],
    on=["prompt_id", "seed"],
    how="inner",
    validate="one_to_one"
)

df = df.merge(
    geneval[
        ["prompt_id", "seed", "correct",
         "geneval_soft_score", "tag"]
    ],
    on=["prompt_id", "seed"],
    how="inner",
    validate="one_to_one"
)

assert len(df) == 3840, (
    f"Expected 3840 matched images, got {len(df)}"
)

assert df["local_seed_idx"].equals(
    df.sort_values(["prompt_id", "seed"])
      .groupby("prompt_id").cumcount()
      .reset_index(drop=True)
), "Local seed ordering mismatch"

assert df.groupby("prompt_id").size().eq(32).all()

print("\nAll metrics successfully merged.")
print("Unique prompts:", df["prompt_id"].nunique())
print("Images per prompt:", sorted(df.groupby("prompt_id").size().unique()))

# --------------------------------------------------
# 3. Select winners under the same candidate groups
# --------------------------------------------------

N_VALUES = [1, 2, 4, 8, 16, 32]

selection_rows = []
group_rows = []

for N in N_VALUES:

    print(f"\nEvaluating N={N}...")

    current = df.copy()

    # Same non-overlapping candidate groups as BoN
    current["group"] = current["local_seed_idx"] // N

    for (group_id, prompt_id), candidates in current.groupby(
        ["group", "prompt_id"]
    ):

        # ImageReward winner
        ir_winner = candidates.loc[
            candidates["image_reward"].idxmax()
        ].copy()

        # CLIP winner
        clip_winner = candidates.loc[
            candidates["clip_cosine_similarity"].idxmax()
        ].copy()

        for verifier, winner in [
            ("ImageReward", ir_winner),
            ("CLIP", clip_winner)
        ]:

            row = winner.to_dict()
            row["N"] = N
            row["group"] = group_id
            row["verifier"] = verifier

            selection_rows.append(row)

    # --------------------------------------------------
    # 4. Calculate group-level evaluation metrics
    # --------------------------------------------------

    selected = pd.DataFrame([
        r for r in selection_rows if r["N"] == N
    ])

    for (verifier, group_id), group_data in selected.groupby(
        ["verifier", "group"]
    ):

        # GenEval overall: average category accuracy
        geneval_by_tag = (
            group_data.groupby("tag")["correct"].mean()
        )

        geneval_overall = geneval_by_tag.mean()

        group_rows.append({
            "N": N,
            "compute_C": N * 64,
            "group": group_id,
            "verifier": verifier,
            "num_prompts": group_data["prompt_id"].nunique(),
            "image_reward_mean": group_data["image_reward"].mean(),
            "clip_similarity_mean": (
                group_data["clip_cosine_similarity"].mean()
            ),
            "geneval_mean": geneval_overall,
            "geneval_soft_score_mean": (
                group_data["geneval_soft_score"].mean()
            )
        })

# --------------------------------------------------
# 5. Save selected images and group metrics
# --------------------------------------------------

selections = pd.DataFrame(selection_rows)
group_metrics = pd.DataFrame(group_rows)

selection_file = os.path.join(
    RESULTS, "verifier_ablation_selections.csv"
)

group_file = os.path.join(
    RESULTS, "verifier_ablation_group_metrics.csv"
)

selections.to_csv(selection_file, index=False)
group_metrics.to_csv(group_file, index=False)

# --------------------------------------------------
# 6. Aggregate results
# --------------------------------------------------

summary = (
    group_metrics.groupby(["N", "compute_C", "verifier"])
    .agg(
        num_groups=("group", "nunique"),
        image_reward_mean=("image_reward_mean", "mean"),
        image_reward_std=("image_reward_mean", "std"),
        clip_similarity_mean=("clip_similarity_mean", "mean"),
        clip_similarity_std=("clip_similarity_mean", "std"),
        geneval_mean=("geneval_mean", "mean"),
        geneval_std=("geneval_mean", "std"),
        geneval_soft_score_mean=(
            "geneval_soft_score_mean", "mean"
        )
    )
    .reset_index()
)

summary_file = os.path.join(
    RESULTS, "verifier_ablation_summary.csv"
)

summary.to_csv(summary_file, index=False)

# --------------------------------------------------
# 7. Paired group-level comparison
# --------------------------------------------------

metric_columns = [
    "image_reward_mean",
    "clip_similarity_mean",
    "geneval_mean",
    "geneval_soft_score_mean"
]

paired_rows = []

for N in N_VALUES:

    subset = group_metrics[group_metrics["N"] == N]

    ir = (
        subset[subset["verifier"] == "ImageReward"]
        .set_index("group")
    )

    cl = (
        subset[subset["verifier"] == "CLIP"]
        .set_index("group")
    )

    for metric in metric_columns:

        differences = (
            cl[metric] - ir[metric]
        ).dropna()

        paired_rows.append({
            "N": N,
            "compute_C": N * 64,
            "metric": metric,
            "mean_CLIP_minus_ImageReward": differences.mean(),
            "std_paired_difference": differences.std(),
            "num_paired_groups": len(differences)
        })

paired = pd.DataFrame(paired_rows)

paired_file = os.path.join(
    RESULTS, "verifier_ablation_paired_differences.csv"
)

paired.to_csv(paired_file, index=False)

# --------------------------------------------------
# 8. Create comparison plots
# --------------------------------------------------

plot_metrics = [
    (
        "geneval_mean",
        "GenEval accuracy",
        "verifier_ablation_geneval.png"
    ),
    (
        "image_reward_mean",
        "ImageReward score",
        "verifier_ablation_imagereward.png"
    )
]

for metric, ylabel, filename in plot_metrics:

    fig, ax = plt.subplots(figsize=(8, 5))

    for verifier in ["ImageReward", "CLIP"]:

        data = summary[
            summary["verifier"] == verifier
        ].sort_values("compute_C")

        ax.plot(
            data["compute_C"],
            data[metric],
            marker="o",
            linewidth=2,
            label=verifier
        )

    ax.set_xlabel("Compute C (trajectory-steps per prompt)")
    ax.set_ylabel(ylabel)
    ax.set_title(f"Final-candidate verifier comparison: {ylabel}")
    ax.set_xscale("log", base=2)
    ax.set_xticks([64, 128, 256, 512, 1024])
    ax.set_xticklabels(["64", "128", "256", "512", "1024"])
    ax.grid(True, alpha=0.3)
    ax.legend()
    fig.tight_layout()

    fig.savefig(
        os.path.join(RESULTS, filename),
        dpi=200
    )

    plt.show()
    plt.close(fig)

# --------------------------------------------------
# 9. Print results
# --------------------------------------------------

print("\n" + "=" * 65)
print("VERIFIER ABLATION COMPLETED")
print("=" * 65)

print("\nSummary:")
print(
    summary[
        [
            "N",
            "compute_C",
            "verifier",
            "image_reward_mean",
            "clip_similarity_mean",
            "geneval_mean",
            "geneval_soft_score_mean",
            "num_groups"
        ]
    ].to_string(index=False)
)

print("\nPaired differences (CLIP - ImageReward):")
print(paired.to_string(index=False))

print("\nOutput files:")
for filename in [
    "verifier_ablation_selections.csv",
    "verifier_ablation_group_metrics.csv",
    "verifier_ablation_summary.csv",
    "verifier_ablation_paired_differences.csv",
    "verifier_ablation_geneval.png",
    "verifier_ablation_imagereward.png"
]:
    print(os.path.join(RESULTS, filename))
```

```python
import os
import pandas as pd

REPO = "/content/inference-scaling-seed-search"
RESULTS = os.path.join(REPO, "results")

selected = pd.read_csv(
    os.path.join(RESULTS, "verifier_ablation_selections.csv")
)

# 1. Per-tag GenEval performance
per_tag = (
    selected.groupby(["N", "verifier", "tag"])
    .agg(
        geneval_accuracy=("correct", "mean"),
        soft_score=("geneval_soft_score", "mean"),
        num_images=("seed", "count")
    )
    .reset_index()
)

per_tag.to_csv(
    os.path.join(RESULTS, "verifier_ablation_per_tag.csv"),
    index=False
)

print("=" * 65)
print("1. PER-TAG GENEVAL RESULTS")
print("=" * 65)

print(
    per_tag.pivot_table(
        index=["N", "tag"],
        columns="verifier",
        values="geneval_accuracy"
    ).round(4).to_string()
)

# 2. Selection overlap
ir = (
    selected[selected["verifier"] == "ImageReward"]
    [["N", "group", "prompt_id", "seed"]]
    .rename(columns={"seed": "ir_seed"})
)

cl = (
    selected[selected["verifier"] == "CLIP"]
    [["N", "group", "prompt_id", "seed"]]
    .rename(columns={"seed": "clip_seed"})
)

overlap = ir.merge(
    cl,
    on=["N", "group", "prompt_id"],
    validate="one_to_one"
)

overlap["same_selected_image"] = (
    overlap["ir_seed"] == overlap["clip_seed"]
)

overlap_summary = (
    overlap.groupby("N")
    .agg(
        total_selections=("same_selected_image", "size"),
        same_selections=("same_selected_image", "sum"),
        overlap_rate=("same_selected_image", "mean")
    )
    .reset_index()
)

overlap_summary.to_csv(
    os.path.join(RESULTS, "verifier_ablation_overlap.csv"),
    index=False
)

print("\n" + "=" * 65)
print("2. SELECTION OVERLAP")
print("=" * 65)

print(overlap_summary.to_string(index=False))

print("\nAnalysis files saved.")
```
